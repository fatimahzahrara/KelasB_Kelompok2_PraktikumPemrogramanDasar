```python
# Meminta pengguna memasukkan nama dan umur
nama = input("Masukan Nama Anda: " )
umur = int(input("Masukkan umur: "))

# Menampilkan sapaan personal
print(f"Halo, {nama}! Umur kamu adalah {umur} tahun.")

# Menghitung tahun lahir pengguna (dari umur)
tahun_sekarang = 2026
tahun_lahir = tahun_sekarang - umur
print(f"Perkiraan tahun lahir kamu adalah {tahun_lahir}.")
```
